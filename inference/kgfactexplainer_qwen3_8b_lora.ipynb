```python
import torch
import json
import re

# The standard Qwen3 ChatML format
qwen3_prompt = """<|im_start|>system
You are a helpful assistant that reasons step-by-step.<|im_end|>
<|im_start|>user
{instruction}

{input}<|im_end|>
<|im_start|>assistant
<|think|>
"""

@torch.no_grad()
def alpaca_inference(
    model,
    tokenizer,
    instruction: str,
    input_text: str,
    temperature: float = 0.1
):

    prompt = qwen3_prompt.format(instruction=instruction, input=input_text)

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        add_special_tokens=False
    ).to(model.device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=4096,
        temperature=0.1,
        top_p=0.95,
        do_sample=(temperature > 0),
        repetition_penalty=1.15,      # Prevents infinite loops/digits
        # no_repeat_ngram_size=4,
        eos_token_id=tokenizer.eos_token_id,
        pad_token_id=tokenizer.eos_token_id,
        use_cache=True,
    )

    # 1. Decode the full sequence
    full_text = tokenizer.decode(outputs[0], skip_special_tokens=False)
    return full_text

    # # 2. Extract everything after the assistant start tag
    # # This removes the System and User blocks
    # if "<|im_start|>assistant" in full_text:
    #     assistant_part = full_text.split("<|im_start|>assistant")[-1]
    # else:
    #     assistant_part = full_text

    # # 3. Remove the thinking process (everything inside <|think|>)
    # # We look for the closing </think> tag
    # if "</think>" in assistant_part:
    #     final_response = assistant_part.split("</think>")[-1]
    # else:
    #     final_response = assistant_part

    # # 4. Clean up any remaining ChatML tags (like <|im_end|>)
    # final_response = final_response.replace("<|im_end|>", "").strip()

    # return final_response

```
