```python
from datasets import load_dataset

ds = load_dataset("nkwbtb/SummaCoz")
```

```python
import pandas as pd

data = ds['train']
data_df = pd.DataFrame(data)

data_df = data_df[(data_df['reason'] != '') & (data_df['origin'] == 'cnndm')]
data_df.head()


# save the dataframe to a CSV file
data_df.to_csv('data/summcoz_train_cnndm.csv', index=False)
```

```python
len(data_df)
```

```python
new_df = pd.DataFrame()

for idx, row in data_df.iterrows():
    if idx % 3 == 0:
        new_df = pd.concat([new_df, row.to_frame().T], ignore_index=True)
```

```python
new_df[90:100]
```

```python
filtered_data_df = new_df[:100]

# save the filtered dataframe to a new CSV file
filtered_data_df.to_csv('data/hallucination_test/summcoz_train_sample_100.csv')
```
