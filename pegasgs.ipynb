```python
from transformers import PegasusTokenizer, PegasusForConditionalGeneration

model_name = "google/pegasus-xsum"
tokenizer_sm = PegasusTokenizer.from_pretrained(model_name)
model_sm = PegasusForConditionalGeneration.from_pretrained(model_name)
```

```python
# save the model and tokenizer
model_sm.save_pretrained("./model/pegasus_xsum_model")
tokenizer_sm.save_pretrained("./model/pegasus_xsum_tokenizer")
```
